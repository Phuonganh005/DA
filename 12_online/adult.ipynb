### 17. (6đ) Độ lệch chuẩn và giá trị trung bình của độ tuổi đối với những người kiếm được hơn 50K mỗi năm (*native-country* feature) và những người kiếm được ít hơn 50K mỗi năm là bao nhiêu?

### 18. (6đ) Có đúng là những người kiếm được hơn 50K đều có trình độ học vấn ít nhất là  high school? (education feature – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate)

```python
df_gt50k = df[df['salary'] == ' >50K']

# Định nghĩa các trình độ học vấn được coi là 'ít nhất high school' theo yêu cầu
high_education_levels = [' Bachelors', ' Prof-school', ' Assoc-acdm', ' Assoc-voc', ' Masters', ' Doctorate']

# Kiểm tra xem TẤT CẢ các trình độ học vấn trong nhóm >50K có nằm trong danh sách đã định nghĩa không
all_high_education = df_gt50k['education'].isin(high_education_levels).all()

if all_high_education:
    print("Đúng vậy, tất cả những người kiếm được hơn 50K đều có trình độ học vấn ít nhất là high school (theo định nghĩa đã cho).")
else:
    # Tìm các trình độ học vấn không nằm trong danh sách đã định nghĩa cho những người kiếm >50K
    low_education_gt50k = df_gt50k[~df_gt50k['education'].isin(high_education_levels)]
    unique_low_education = low_education_gt50k['education'].unique()

    print("Không. Có những người kiếm được hơn 50K nhưng có trình độ học vấn KHÔNG nằm trong danh sách 'ít nhất là high school' đã định nghĩa.")
    print(f"Các trình độ học vấn không đạt yêu cầu tìm thấy trong nhóm >50K: {unique_low_education}")
```

### 20. (6đ) Tỉ lệ người married hay single men (marital-status feature) chiếm nhiều hơn trong số những người kiếm được >50k? Những người có tình trạng hôn nhân (married) có marital-status feature bắt đầu với Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), còn lại được coi là single men.

```python
df_gt50k_men = df[(df['salary'] == ' >50K') & (df['sex'] == ' Male')]

# Định nghĩa các trạng thái hôn nhân được coi là 'Married'
married_status = [' Married-civ-spouse', ' Married-spouse-absent', ' Married-AF-spouse']

# Phân loại đàn ông thành 'Married' hoặc 'Single'
df_gt50k_men['marital_status_group'] = df_gt50k_men['marital-status'].apply(
    lambda x: 'Married' if x in married_status else 'Single'
)

# Đếm số lượng của mỗi nhóm
marital_counts = df_gt50k_men['marital_status_group'].value_counts()

display(marital_counts)

if marital_counts.get('Married', 0) > marital_counts.get('Single', 0):
    print("Trong số những người đàn ông kiếm được hơn 50K, số người đã kết hôn (Married) chiếm tỷ lệ nhiều hơn.")
elif marital_counts.get('Single', 0) > marital_counts.get('Married', 0):
    print("Trong số những người đàn ông kiếm được hơn 50K, số người độc thân (Single) chiếm tỷ lệ nhiều hơn.")
else:
    print("Trong số những người đàn ông kiếm được hơn 50K, số người đã kết hôn và độc thân có tỷ lệ tương đương.")
```

### 21. (6đ) Số giờ tối đa một người làm việc mỗi tuần (hours-per-week feature) là bao nhiêu? Có bao nhiêu người làm việc với số giờ như vậy và tỷ lệ những người kiếm được nhiều (>50K) trong số đó là bao nhiêu?

```python
max_hours_per_week = df['hours-per-week'].max()
print(f"Số giờ tối đa một người làm việc mỗi tuần là: {max_hours_per_week}")

# Lọc những người làm việc với số giờ tối đa
max_hours_workers = df[df['hours-per-week'] == max_hours_per_week]
num_max_hours_workers = max_hours_workers.shape[0]
print(f"Có {num_max_hours_workers} người làm việc với số giờ tối đa này.")

# Đếm số người kiếm được >50K trong số những người làm việc tối đa
num_gt50k_max_hours_workers = max_hours_workers[max_hours_workers['salary'] == ' >50K'].shape[0]

# Tính tỷ lệ phần trăm
percentage_gt50k_max_hours_workers = (num_gt50k_max_hours_workers / num_max_hours_workers) * 100
print(f"Tỷ lệ những người kiếm được nhiều (>50K) trong số đó là: {percentage_gt50k_max_hours_workers:.2f}%")
```
